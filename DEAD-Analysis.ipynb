```python
emp_df['acs_end_year'] = pd.to_datetime(emp_df['data_collection_end_date']).dt.year
age_sex_rows = emp_df[
    emp_df['sex'].isin(['Male', 'Female']) &
    emp_df['age_range'].notna() &
    emp_df['is_in_labor_force'].isna() &
    emp_df['is_in_armed_forces'].isna() &
    emp_df['employment_status'].isna()
]

# total 16+ population per county-year 
total_16plus = (
    age_sex_rows.groupby(['county_fips_code', 'acs_end_year'])['population']
    .sum().reset_index(name='total_16plus_pop')
)

# young adult population 
young_bins = ['20 and 21 Years', '22 to 24 Years', '25 to 29 Years', '30 to 34 Years']

young_pop = (
    age_sex_rows[age_sex_rows['age_range'].isin(young_bins)]
    .groupby(['county_fips_code', 'acs_end_year'])['population']
    .sum().reset_index(name='young_adult_pop')
)

age_share = young_pop.merge(total_16plus, on=['county_fips_code', 'acs_end_year'])
age_share['pct_young_adult'] = age_share['young_adult_pop'] / age_share['total_16plus_pop']

```

model

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor

county_month['log_liters_per_adult'] = np.log(county_month['total_liters'] / county_month['adult_population'])
county_month['log_income'] = np.log(county_month['median_household_income'])
county_month['log_density'] = np.log(county_month['pop_density'])
county_month['stores_per_10k'] = county_month['n_stores'] / county_month['adult_population'] * 10000
county_month['log_stores_per_10k'] = np.log(county_month['stores_per_10k'].replace(0, np.nan))  # guard zero-store counties

# urban indicator — median split on density 
county_month['urban'] = (county_month['pop_density'] > county_month['pop_density'].median()).astype(int)

county_month['month'] = county_month['month'].astype(int)

model = smf.wls(
    formula='''log_liters_per_adult ~ log_income + pct_young_adult + employment_rate 
               + log_density + log_stores_per_10k 
               + C(month) * urban''',
    data=county_month,
    weights=county_month['adult_population']
)

results = model.fit()
print(results.summary())

results_cl = model.fit(cov_type='cluster', cov_kwds={'groups': county_month['county_fips_code']})
print(results_cl.summary())

X = county_month[['log_income', 'pct_young_adult', 'employment_rate', 'log_density', 'log_stores_per_10k']].dropna()
X = sm.add_constant(X)

vif = pd.DataFrame()
vif['variable'] = X.columns
vif['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print(vif)
```

```python
conf_int = results_cl.conf_int()
conf_int.columns = ['ci_lower', 'ci_upper']
summary_table = pd.concat([results_cl.params, conf_int], axis=1)
summary_table.columns = ['coef', 'ci_lower', 'ci_upper']
print(summary_table)

county_month['predicted'] = results_cl.predict(county_month)
county_month['residual'] = county_month['log_liters_per_adult'] - county_month['predicted']

county_avg_residual = (
    county_month.groupby('county_fips_code')['residual']
    .mean().sort_values()
    .reset_index()
)
print(county_avg_residual.head(10))   # most under-predicted (buying less than expected)
print(county_avg_residual.tail(10))   # most over-predicted (buying more than expected)

import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12,4))
ax[0].scatter(results_cl.fittedvalues, results_cl.resid)
ax[0].set_xlabel('Fitted'); ax[0].set_ylabel('Residual')

sm.qqplot(results_cl.resid, line='45', ax=ax[1])
plt.show()
```

This is a good, interpretable result. Let me walk through what actually changed and what it means.

## The headline finding: cluster-robust SEs completely changed your conclusions

This is exactly the effect I flagged earlier, and now you can see it directly. Compare the two tables:

| Variable | Nonrobust SE | Cluster SE | Nonrobust p | Cluster p |
|---|---|---|---|---|
| `log_income` | 0.048 | 0.332 | <0.001 | 0.016 |
| `pct_young_adult` | 0.113 | 0.541 | <0.001 | 0.212 |
| `employment_rate` | 0.456 | 2.785 | <0.001 | 0.236 |

The cluster-robust standard error on `log_income` is **7x larger** than the naive one. For `pct_young_adult` and `employment_rate`, the coefficients didn't change at all (they can't — clustering only changes the SEs, not the point estimates), but they went from "highly significant" to **not statistically distinguishable from zero**. This is precisely because those variables come from an annual ACS table and don't actually vary month-to-month within a county-year — you effectively have ~99 independent observations informing those effects, not 4,752, and the naive model was fooling itself into false precision. This is genuinely one of your more important findings to report honestly: **age structure and employment rate showed strong associations in the naive model, but that association disappears once you correctly account for the repeated-measures structure of the data.**

## What survives clustering — your real findings

**`log_income` = -0.80** (still significant, p=0.016): a 1% increase in median household income is associated with roughly a 0.8% *decrease* in liters per adult, holding other factors fixed. Higher-income counties buy less alcohol per adult in this model.

**`log_density` = 0.27** (highly significant, p<0.001): a 1% increase in population density is associated with a 0.27% increase in liters per adult.

**`log_stores_per_10k` = 0.49** (highly significant, p=0.001): a 1% increase in stores per capita is associated with roughly a 0.49% increase in liters per adult — your strongest access-related finding, and a sensible one (retail availability tracks purchasing).

**`urban` = 0.225** (significant, p=0.007): since this is a coefficient on the log outcome, exponentiate it — `exp(0.225) - 1 ≈ 25%` — urban counties show about 25% higher baseline liters-per-adult than rural counties in January, before accounting for the seasonal interactions.

## Seasonal pattern (month dummies, relative to January)

Purchasing rises steadily from spring through the end of the year, peaking around **May (+17%)** and **December (+16%)**, with most months from April onward showing statistically significant increases over January. This is your "overview of seasonal purchasing patterns" deliverable, essentially ready to write up.

## Month × urban interactions

Several are significant — February, March, April, August, and October show a notably larger seasonal boost in urban counties than rural ones relative to January (e.g., March: urban counties see an *additional* 12% jump beyond the general seasonal effect). This tells you **seasonal purchasing patterns genuinely differ by community type**, which is exactly the "whether these patterns differ across types of communities" question the proposal set out to answer.

## VIF — no multicollinearity problem

All five VIFs are under 2.2 (the common rule of thumb for concern is 5, sometimes 10). You can trust these coefficients individually without worrying that they're contaminated by predictors moving together. Good news, and worth stating plainly in your write-up rather than hedging.

## Two things worth flagging in limitations

**Durbin-Watson = 0.50** is quite low (2.0 = no autocorrelation, 0 = strong positive autocorrelation). This means residuals are correlated month-to-month within a county — sales in March look like sales in April for the same county, more than a well-specified model would predict. Cluster-robust SEs by county partially account for this (that's exactly what clustering is designed to fix), but it's worth naming as a limitation: the model doesn't explicitly capture within-county trends over time (no lagged term, no county fixed effects), and some of what's showing up as a strong seasonal or urban effect could partly reflect persistent county-level habits not fully captured by your predictors.

**Residual normality** — the Jarque-Bera test rejects normality (p ≈ 2×10⁻²⁷), and skew/kurtosis are mildly off from a normal distribution. With nearly 5,000 observations this is common and not fatal to inference (your standard errors and CIs are still valid via cluster-robust estimation and the sample size), but worth a quick residual plot to confirm there's no dramatic pattern (like a few extreme outlier counties dragging the tail).

## Bottom line for your report

R² = 0.595 is solid for a model of this kind — you're explaining about 60% of the variation in monthly per-adult liters purchased. Your defensible, statistically supported findings are: **income (negative), density (positive), store access (positive), urban status (positive), a clear seasonal pattern peaking in spring and December, and seasonal patterns that differ meaningfully between urban and rural counties.** Age structure and employment rate should be reported as **not statistically significant once properly accounting for clustering** — which is itself an honest and useful finding, not a failure of the model.
