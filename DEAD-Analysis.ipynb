```python
df1 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0001.csv", dtype={'employment_status': str}, low_memory=False)
df2 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0002.csv", dtype={'employment_status': str}, low_memory=False)
df3 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0003.csv", dtype={'employment_status': str}, low_memory=False)
df4 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0004.csv", dtype={'employment_status': str}, low_memory=False)
df5 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0005.csv", dtype={'employment_status': str}, low_memory=False)
df6 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0006.csv", dtype={'employment_status': str}, low_memory=False)
df7 = pd.read_csv("iowa_population_16_years_and_over_by_sex_age_and_employment_status_acs_5_year_estimates_536_rows_part_0007.csv", dtype={'employment_status': str}, low_memory=False)


emp_df = pd.concat([df1, df2, df3, df4, df5, df6, df7], ignore_index=True)
```

```python
emp_df = emp_df[emp_df['geography_type'].str.strip().str.lower() == 'county']
emp_df.info()
```

```python
for col in ['geography_type', 'sex', 'age_range', 'is_in_labor_force', 
            'is_in_armed_forces', 'employment_status', 'variable_description', 'data_collection_period']:
    print(col, '→', emp_df[col].unique())
    print()
```

```python
emp_df['acs_end_year'] = pd.to_datetime(emp_df['data_collection_end_date']).dt.year
age_sex_rows = emp_df[
    emp_df['sex'].isin(['Male', 'Female']) &
    emp_df['age_range'].notna() &
    emp_df['is_in_labor_force'].isna() &
    emp_df['is_in_armed_forces'].isna() &
    emp_df['employment_status'].isna()
]

# total 16+ population per county-year 
total_16plus = (
    age_sex_rows.groupby(['county_fips_code', 'acs_end_year'])['population']
    .sum().reset_index(name='total_16plus_pop')
)

# young adult population 
young_bins = ['20 and 21 Years', '22 to 24 Years', '25 to 29 Years', '30 to 34 Years']

young_pop = (
    age_sex_rows[age_sex_rows['age_range'].isin(young_bins)]
    .groupby(['county_fips_code', 'acs_end_year'])['population']
    .sum().reset_index(name='young_adult_pop')
)

age_share = young_pop.merge(total_16plus, on=['county_fips_code', 'acs_end_year'])
age_share['pct_young_adult'] = age_share['young_adult_pop'] / age_share['total_16plus_pop']

```

```python
age_share
```

```python
civilian_labor = emp_df[
    (emp_df['is_in_labor_force'] == True) &
    (emp_df['is_in_armed_forces'] == False) &
    (emp_df['employment_status'].isin(['Employed', 'Unemployed']))
]

emp_by_status = (
    civilian_labor.groupby(['county_fips_code', 'acs_end_year', 'employment_status'])['population']
    .sum().unstack('employment_status').reset_index()
)

emp_by_status['employment_rate'] = (
    emp_by_status['Employed'] / (emp_by_status['Employed'] + emp_by_status['Unemployed'])
)

emp_by_status
```

```python

age_share_latest = age_share[age_share['acs_end_year'] == 2024][['county_fips_code', 'pct_young_adult']]
emp_latest = emp_by_status[emp_by_status['acs_end_year'] == 2024][['county_fips_code', 'employment_rate']]

county_month = county_month.merge(age_share_latest, on='county_fips_code', how='left')
county_month = county_month.merge(emp_latest, on='county_fips_code', how='left')
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor

county_month['log_liters_per_adult'] = np.log(county_month['total_liters'] / county_month['adult_population'])
county_month['log_income'] = np.log(county_month['median_household_income'])
county_month['log_density'] = np.log(county_month['pop_density'])
county_month['stores_per_10k'] = county_month['n_stores'] / county_month['adult_population'] * 10000
county_month['log_stores_per_10k'] = np.log(county_month['stores_per_10k'].replace(0, np.nan))  # guard zero-store counties

# urban indicator — median split on density 
county_month['urban'] = (county_month['pop_density'] > county_month['pop_density'].median()).astype(int)

county_month['month'] = county_month['month'].astype(int)

model = smf.wls(
    formula='''log_liters_per_adult ~ log_income + pct_young_adult + employment_rate 
               + log_density + log_stores_per_10k 
               + C(month) * urban''',
    data=county_month,
    weights=county_month['adult_population']
)

results = model.fit()
print(results.summary())

results_cl = model.fit(cov_type='cluster', cov_kwds={'groups': county_month['county_fips_code']})
print(results_cl.summary())

X = county_month[['log_income', 'pct_young_adult', 'employment_rate', 'log_density', 'log_stores_per_10k']].dropna()
X = sm.add_constant(X)

vif = pd.DataFrame()
vif['variable'] = X.columns
vif['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print(vif)
```
